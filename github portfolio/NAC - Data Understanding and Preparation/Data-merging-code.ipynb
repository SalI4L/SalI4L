```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
from scipy import stats
import sklearn
import missingno as msno
```

```python

def read_data_from_folders(root_folder):
    # Initialize an empty list to store DataFrames
    dfs = []

    # Traverse through all folders in the specified directory
    for subdir, dirs, files in os.walk(root_folder):
        for file in files:
            # Assuming all files are Excel files, adjust the condition based on your data format
            if file.endswith(".xlsx") or file.endswith(".xls"):
                file_path = os.path.join(subdir, file)

                # Read data from each Excel file
                data = pd.read_excel(file_path)

                # Append data to the list of DataFrames
                dfs.append(data)

    # Concatenate all DataFrames in the list into a single DataFrame
    combined_dataframe = pd.concat(dfs, ignore_index=True)

    return combined_dataframe

# Replace with your actual root folder path
root_folder = "C:/Users/sally/Desktop/School/Year 1/2023-24b-fai1-adsai-SallyIbrahim211066/data/2022_2023"


# Read data from folders
combined_dataframe = read_data_from_folders(root_folder)

# Display the combined DataFrame
combined_dataframe.head()

# Save the combined DataFrame to a CSV file
combined_dataframe.to_csv('combined_data.csv', index=False)
```
